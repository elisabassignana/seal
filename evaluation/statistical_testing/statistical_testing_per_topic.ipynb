### Prepare data with matched topics
We choose 2 topics, for which the distribution of upper, middle and low social class user queries is approximately balanced.

```python
import pandas as pd

df = pd.read_csv("../all_metrics/all_metrics_topic_0_3.csv")

df
```

```python
df["topic_label"].value_counts()

df["user"] = df["id"].str.split("p").str[0]

df_business = df[df["topic_label"] == "Business Writing & Analysis"]
df_travel = df[df["topic_label"] == "Travel & Diet/Meal Planning"]
```

```python
df_travel = (
    df_travel.groupby("user", as_index=False)
      .agg({
          "social_class": "first",
          **{col: "mean" for col in df.select_dtypes(include="number").columns}
      })
)

df_travel
```

```python
df_business = (
    df_business.groupby("user", as_index=False)
      .agg({
          "social_class": "first",
          **{col: "mean" for col in df.select_dtypes(include="number").columns}
      })
)

df_business
```

```python
# complexity measures
complexity = ["average_age_of_acquisition",
              "syllables_per_word",
              "adversative_connectives",
              "negations_density"] 


# linguistic features
linguistic = ["avg_valence", 
        "avg_arousal",
        "avg_dominance", 
        "avg_intensity_anger", 
        "avg_intensity_anticipation", 
        "avg_intensity_disgust", 
        "avg_intensity_fear", 
        "avg_intensity_joy", 
        "avg_intensity_sadness", 
        "avg_intensity_surprise", 
        "avg_intensity_trust", 
        "sentiment_score"] 

# all other measures
others = ["normalized_helpfulness_responses",
          'humt_responses',
          'sociot_status_responses',
          'sociot_social_distance_responses',
          'sociot_gender_responses',
          'sociot_warmth_responses',
          'syco_framing',
          'syco_validation',
          'syco_indirectness']

models = ["gemma-4-31B", "llama-3.3-70B", "qwen-3.5-27B", "gpt-5.5"]
ling = [f"ling_{l}_{m}" for l in linguistic for m in models]
cplx = [f"cplx_{c}_{m}" for c in complexity for m in models]
oth = [f"{o}_{m}" for o in others for m in models]

measures = ling + cplx + oth

measures
```

### Kruskal-Wallis test per topic
We check whether the score distribution for each social class originates from the same population (i.e., is there a **significant difference between the 3 classes**?). Correct for the number of test (i.e., measures) per model using Benjamin-Hochberg correction.

```python
import pandas as pd
import numpy as np
from scipy import stats
import scikit_posthocs as sp
from statsmodels.stats.multitest import multipletests

# set desired topic
TOPIC = "travel" # business

dfs = {"business": df_business, "travel": df_travel}
df = dfs[TOPIC]

order = ['low', 'middle', 'upper']

def get_model(col, models):
    for m in models:
        if col.endswith(m):
            return m
    return None

rows = []
for measure in measures:
    model = get_model(measure, models)
    if model is None:
        continue

    sub = df.dropna(subset=[measure])
    groups = [sub.loc[sub['social_class'] == lvl, measure] for lvl in order]

    h_stat, p_omnibus = stats.kruskal(*groups)
    n = sum(len(g) for g in groups)
    eps_sq = h_stat / ((n**2 - 1) / (n + 1))

    rows.append({
        'model': model,
        'measure': measure,
        'p_omnibus': p_omnibus,
        'epsilon_squared': eps_sq,

    })

results = pd.DataFrame(rows)

results['p_omnibus_fdr'] = np.nan
results['significant_fdr'] = False

for model in models:
    mask = results['model'] == model
    pvals = results.loc[mask, 'p_omnibus'].values
    if len(pvals) == 0:
        continue
    reject, p_adj, _, _ = multipletests(pvals, alpha=0.05, method='fdr_bh')
    results.loc[mask, 'p_omnibus_fdr'] = p_adj
    results.loc[mask, 'significant_fdr'] = reject

results = results[['model', 'measure', 'p_omnibus', 'p_omnibus_fdr', 'significant_fdr', 'epsilon_squared']]
results = results.sort_values(['model', 'p_omnibus_fdr'])

results
```

```python
results.to_csv("all_group_differences_travel.csv", index=False)
```

## Analysis of results for topic "Travel & Diet/Meal Planning"

### All groups
For each model, show for which measures there is a significant difference between the **three social classes** (p < 0.05).

```python
travel = pd.read_csv("all_group_differences_travel.csv")

results = travel[['model', 'measure', 'p_omnibus', 'p_omnibus_fdr', 'significant_fdr', 'epsilon_squared']]
results = results.sort_values(['model', 'p_omnibus_fdr'])

# ---- overall summary ----
n_sig = results['significant_fdr'].sum()
n_total = len(results)
print(f"Significant tests (FDR-corrected): {n_sig}/{n_total}\n")

summary = results.groupby('model')['significant_fdr'].agg(['sum', 'count'])
summary.columns = ['n_significant', 'n_total']
print(summary)

# ---- significant measures with p-values, by model ----
print("\nSignificant measures (FDR-corrected), by model:")
for model in models:
    model_sig = results[(results['model'] == model) & (results['significant_fdr'])].sort_values('p_omnibus_fdr')
    print(f"\n  {model}:")
    if model_sig.empty:
        print("    none")
    else:
        print(model_sig[['measure', 'p_omnibus', 'p_omnibus_fdr', 'epsilon_squared']]
              .to_string(index=False))

#results
```
