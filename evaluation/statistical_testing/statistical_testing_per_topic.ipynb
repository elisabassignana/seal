### Kruskal-Wallis test per topic
We check whether the score distribution for each social class originates from the same population (i.e., is there a **significant difference between the 3 classes**?). Posthoc, we perform Dunn's test for pairwise comparisons of social classes and use Benjamin Hochberg correction for multiple tests.

```python
import pandas as pd
import numpy as np
from scipy import stats
import scikit_posthocs as sp
from statsmodels.stats.multitest import multipletests

order = ['low', 'middle', 'upper']
measures = cols_rel

def cliffs_delta(x, y):
    x, y = np.array(x), np.array(y)
    diff = x[:, None] - y[None, :]
    return (np.sum(diff > 0) - np.sum(diff < 0)) / (len(x) * len(y))

rows = []
for measure in measures:
    sub = df_business.dropna(subset=[measure])  # set to df filtered by topic of choice
    groups = [sub.loc[sub['social_class'] == lvl, measure] for lvl in order]

    h_stat, p_omnibus = stats.kruskal(*groups)
    n = sum(len(g) for g in groups)
    eps_sq = h_stat / ((n**2 - 1) / (n + 1))

    # Dunn's test, Holm-corrected WITHIN this measure's 3 pairs
    dunn_p = sp.posthoc_dunn(sub, val_col=measure, group_col='social_class', p_adjust='holm')

    pairs = [('low', 'middle'), ('low', 'upper'), ('middle', 'upper')]
    for g1, g2 in pairs:
        x = sub.loc[sub['social_class'] == g1, measure]
        y = sub.loc[sub['social_class'] == g2, measure]
        delta = cliffs_delta(x, y)
        direction = g1 if delta > 0 else (g2 if delta < 0 else "tie")

        rows.append({
            'measure': measure,
            'p_omnibus': p_omnibus,
            'epsilon_squared': eps_sq,
            'comparison': f'{g1} vs {g2}',
            'dunn_p_holm': dunn_p.loc[g1, g2],   # significance for this pair
            'cliffs_delta': delta,               # magnitude + direction
            'higher_group': direction,
        })

results = pd.DataFrame(rows)

# --- Flag: is this specific pairwise comparison significant AND meaningful? ---
def flag_pair(row):
    if row['p_omnibus'] >= 0.05:
        return 'omnibus not significant'
    elif row['dunn_p_holm'] >= 0.05:
        return 'no significant difference (pair)'
    else:
        return 'significant difference (pair)'


results['pair_flag'] = results.apply(flag_pair, axis=1)

results['pair_flag'] = results.apply(flag_pair, axis=1)

results = results[['measure', 'p_omnibus', 'epsilon_squared', 
                    'comparison', 'dunn_p_holm', 'cliffs_delta', 'higher_group', 'pair_flag']] 

results
```

```python
results.to_csv("all_group_differences_business.csv", index=False)
```

### Mann-Whitney-U-test per topic
We also test whether there is a significant difference between only the **upper** and **lower** class by using Mann-Whitney-U-test.

```python
import pandas as pd
import numpy as np
from scipy import stats
from statsmodels.stats.multitest import multipletests

measures = cols_rel

def cliffs_delta(x, y):
    x, y = np.array(x), np.array(y)
    diff = x[:, None] - y[None, :]
    return (np.sum(diff > 0) - np.sum(diff < 0)) / (len(x) * len(y))

rows = []
for measure in measures:
    sub = df_business.dropna(subset=[measure])  # set to df filtered by topic of choice
    x = sub.loc[sub['social_class'] == 'upper', measure]
    y = sub.loc[sub['social_class'] == 'low', measure]

    stat, p = stats.mannwhitneyu(x, y, alternative='two-sided')
    delta = cliffs_delta(x, y)
    direction = 'upper' if delta > 0 else ('low' if delta < 0 else 'tie')

    rows.append({
        'measure': measure,
        'p_value': p,
        'cliffs_delta': delta,
        'higher_group': direction,
    })

results = pd.DataFrame(rows)


# set flag
def flag(row):
    if row['p_value'] >= 0.05:
        return 'no significant difference'
    else:
        return "significant difference"
    

results['flag'] = results.apply(flag, axis=1)

results = results[['measure', 'p_value', 'cliffs_delta', 'higher_group', 'flag']] 

display(results)
```

```python
results.to_csv("upper_vs_low_differences_business.csv", index=False)
```

## Analysis of results for topic "Travel & Diet/Meal Planning"

```python
# add in interesting complexity measures
meas = ["cplx_syllables_per_word", "cplx_average_age_of_acquisition", "cplx_adversative_connectives"]
models = ["gemma-4-31B", "llama-3.3-70B", "qwen-3.5-27B", "gpt-5.5"]

comp_cols = [f"{m}_{model}" for m in meas for model in models]
comp_cols
```

### Upper vs lower 

For each model, show for which measures there is a significant difference between the **upper** and **lower** social class (p < 0.05).

```python
res_ul = pd.read_csv("upper_vs_low_differences_travel.csv")
res_ul = res_ul[
    ~res_ul["measure"].str.startswith("cplx", na=False)
    | res_ul["measure"].isin(comp_cols)
]

models = ["gemma-4-31B", "llama-3.3-70B", "qwen-3.5-27B", "gpt-5.5"]

for m in models:
    print(m)
    res_model = res_ul[res_ul["measure"].str.endswith(m)]
    print(len(res_model))
    sign = res_model[res_model["flag"] == "significant difference"]
    display(sign)
    print(len(sign))
    print()
```

### All groups
For each model, show for which measures there is a significant difference between the **three social classes** (p < 0.05).

```python
res_ag = pd.read_csv("all_group_differences_travel.csv")
res_ag = res_ag[
    ~res_ag["measure"].str.startswith("cplx", na=False)
    | res_ag["measure"].isin(comp_cols)
]

models = ["gemma-4-31B", "llama-3.3-70B", "qwen-3.5-27B", "gpt-5.5"]

for m in models:
    print(m)
    res_model = res_ag[res_ag["measure"].str.endswith(m)]
    print(len(res_model))

    sign = res_model[res_model["pair_flag"] != "omnibus not significant"]

    sign = (
        sign[["measure", "p_omnibus"]]
        .drop_duplicates()
        .reset_index(drop=True)
    )

    display(sign)
    print(len(sign))
    print()
```
