### Statistical Testing

```python
import pandas as pd

path = "all_metrics/all_metrics_per_user.csv"

df = pd.read_csv(path, index_col=False)

cols = list(df.columns)
```

```python
exclude = ['user',
 'social_class',
 'gender',
 'gender_other',
 'age',
 'nationality',
 'ethnicity',
 'ethnicity_other',
 'marital',
 'marital_other',
 'language',
 'language_other',
 'religion',
 'religion_other',
 'education',
 'mum_education',
 'dad_education',
 'home',
 'home_other',
 'employment',
 'occupation',
 'mother_occupation',
 'father_occupation',
 'hobbies',
 'hobbies_other',
 'tech',
 'tech_other',
 'know_nlp',
 'know_nlp_other',
 'use_nlp',
 'use_nlp_other',
 'would_nlp',
 'would_nlp_other',
 'frequency_llm',
 'llm_use',
 'llm_other',
 'usecases',
 'usecases_other',
 'contexts',
 'contexts_other',
 'comments',
 'ses']

cols_rel = [col for col in cols if not col in exclude]
cols_rel = [col for col in cols_rel if not col.startswith("std")]
#cols_rel = [col for col in cols_rel if not col.startswith("cplx")] # cplx ling hlepfulness
cols_rel = [col for col in cols_rel if not col.startswith("ling")]
cols_rel = [col for col in cols_rel if not col.startswith("helpfulness")]

complexity = ["average_age_of_acquisition", 
              "conceptual_graph_ontology_number_of_concpets",
              "dale_chall_readability_score",
              "gunning_fog_index",
              "word_concreteness",
              "average_cefr_level",
              "mean_word_length",
              "passive_constructions_density",
              "connectives"]

linguistic = ["avg_valence", 
        "avg_arousal",
        "avg_dominance",
        "avg_intensity_anger",
        "avg_intensity_anticipation",
        "avg_intensity_disgust",
        "avg_intensity_fear",
        "avg_intensity_joy",
        "avg_intensity_sadness",
        "avg_intensity_surprise",
        "avg_intensity_trust",
        "sentiment_score"]

models = ["gemma-4-31B", "llama-3.3-70B", "qwen-3.5-27B", "gpt-5.5"]

ling = [f"ling_{l}_{m}" for l in linguistic for m in models]
#cplx = [f"cplx_{c}_{m}" for c in complexity for m in models]

cols_rel = cols_rel + ling

cols_rel
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import scikit_posthocs as sp
from statsmodels.stats.multitest import multipletests

order = ['low', 'middle', 'upper']
measures = cols_rel

def cliffs_delta(x, y):
    x, y = np.array(x), np.array(y)
    diff = x[:, None] - y[None, :]
    return (np.sum(diff > 0) - np.sum(diff < 0)) / (len(x) * len(y))

rows = []
for measure in measures:
    sub = df.dropna(subset=[measure])
    groups = [sub.loc[sub['social_class'] == lvl, measure] for lvl in order]

    h_stat, p_omnibus = stats.kruskal(*groups)
    n = sum(len(g) for g in groups)
    eps_sq = h_stat / ((n**2 - 1) / (n + 1))

    # Dunn's test, Holm-corrected WITHIN this measure's 3 pairs
    dunn_p = sp.posthoc_dunn(sub, val_col=measure, group_col='social_class', p_adjust='holm')

    pairs = [('low', 'middle'), ('low', 'upper'), ('middle', 'upper')]
    for g1, g2 in pairs:
        x = sub.loc[sub['social_class'] == g1, measure]
        y = sub.loc[sub['social_class'] == g2, measure]
        delta = cliffs_delta(x, y)
        direction = g1 if delta > 0 else (g2 if delta < 0 else "tie")

        rows.append({
            'measure': measure,
            'p_omnibus': p_omnibus,
            'epsilon_squared': eps_sq,
            'comparison': f'{g1} vs {g2}',
            'dunn_p_holm': dunn_p.loc[g1, g2],   # <-- significance for this pair
            'cliffs_delta': delta,               # <-- magnitude + direction
            'higher_group': direction,
        })

results = pd.DataFrame(rows)

# --- Correction 1: omnibus p across all measures ---
#omnibus_unique = results.drop_duplicates('measure')[['measure', 'p_omnibus']]
#_, p_fdr, _, _ = multipletests(omnibus_unique['p_omnibus'], method='fdr_bh')
#omnibus_unique['p_omnibus_fdr'] = p_fdr
#results = results.merge(omnibus_unique[['measure', 'p_omnibus_fdr']], on='measure')

# --- Correction 2: pairwise dunn_p across ALL rows (all measures x all pairs) ---
#_, dunn_p_fdr, _, _ = multipletests(results['dunn_p_holm'], method='fdr_bh')
#results['dunn_p_fdr'] = dunn_p_fdr

# --- Flag: is this specific pairwise comparison significant AND meaningful? ---
def flag_pair(row):
    if row['p_omnibus'] >= 0.05:
        return 'omnibus not significant'
    elif row['dunn_p_holm'] >= 0.05:
        return 'no significant difference (pair)'
    else:
        return 'significant difference (pair)'
    
    #elif abs(row['cliffs_delta']) < 0.147:
    #    return 'significant but negligible'
    #else:
    #    return 'significant and meaningful'

results['pair_flag'] = results.apply(flag_pair, axis=1)

results['pair_flag'] = results.apply(flag_pair, axis=1)

results = results[['measure', 'p_omnibus', 'epsilon_squared', # 'p_omnibus_fdr',
                    'comparison', 'dunn_p_holm', 'cliffs_delta', 'higher_group', 'pair_flag']] # 'dunn_p_fdr',

results
```
